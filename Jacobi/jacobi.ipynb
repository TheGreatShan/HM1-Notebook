## Aufgabe 1

```python
import numpy as np
```

Teilaufgabe a

```python
A = np.array([[8, -2, 1, 0], [1, 10, -3, 2], [-2, 1, 9, -1], [0, 2, -1, 7]])
b = np.array([[12, -13, 14, 3]]).T

print(A)
print(b)
```

```python
# k = -1 entfernt die Diagonalen, damit es auch wirklich die echten utneren und oberen Dreiecksmatrizen sind.
L = np.tril(A, k = -1) # Das gibt die untere Dreiecksmatrix zurück
R = np.triu(A, k = -1) # das gibt die obere Dreiekcsmatrix zurück


# Doppelte Anwendung von Diag nötig, damit die Matrix wieder erstellt wird. Sonst hat man nur noch ein 1D Vektor.
# np.eye wäre eine weitere Option um das Probolem zu lösen.
D = np.diag(np.diag(A))
```

Teilaufgabe b

```python
D_inv = np.linalg.inv(D)
B =  D_inv * (L + R)
c = D_inv * b

print(f"B = \n {B}")
print(f"c = \n {c}")
```

Teilaufgabe c)

Die Diagonaldominanz wir berechnet um zu bestimmen, ob überhaupt eine Konvergenz stattfinden kann um die Berechnung mit Jacobi oder Gauss-Seidel durchzuführen.
